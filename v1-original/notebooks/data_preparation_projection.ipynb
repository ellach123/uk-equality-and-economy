# Data Projection

```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler
```

```python
# Load the dataset
file_path = '../../data/data_selected_for_tableau.csv'
data = pd.read_csv(file_path)
```

```python
data.info()
```

```python
# Keep the geography columns
geography = data['geography']
geography_code = data['geography_code']

# Select the numeric columns
numerical_data = data.select_dtypes(include=['int64'])

```

```python
# Standardise the data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(numerical_data)
```

```python
# Project to two dimensions with PCA
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data)
pca_df = pd.DataFrame(pca_result, columns=['PCA1', 'PCA2'])
```

```python
# Save the PCA result to CSV
pca_combined_df = pd.concat([geography.reset_index(drop=True), pca_df], axis=1)
pca_output_file_path = '../../data/pca_projected_data.csv'
pca_combined_df.to_csv(pca_output_file_path, index=False)
```

```python

# Randomly sample part of the data for t-SNE
sampled_data = scaled_data[:1000]  # sample the first 1000 rows
sampled_geography = geography.iloc[:1000].reset_index(drop=True)
tsne = TSNE(n_components=2, random_state=42, n_iter=300)
tsne_result = tsne.fit_transform(sampled_data)
tsne_df = pd.DataFrame(tsne_result, columns=['TSNE1', 'TSNE2'])
```

```python
# Save the t-SNE result to CSV
tsne_combined_df = pd.concat([sampled_geography, tsne_df], axis=1)
tsne_output_file_path = '../../data/tsne_projected_data.csv'
tsne_combined_df.to_csv(tsne_output_file_path, index=False)
```
